```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from keras.models import Sequential
from keras.layers import Dense
from keras.optimizers import SGD
```

```python
import kagglehub
from kagglehub import KaggleDatasetAdapter

# Set the path to the file you'd like to load
path = kagglehub.dataset_download("yasserh/titanic-dataset")

dataset = pd.read_csv(path+'/Titanic-Dataset.csv')
dataset.head(3)
```

```python
dataset = dataset.drop('Name', axis=1)
dataset = dataset.drop('Ticket', axis=1)
```

```python
dataset.head(3)
```

```python
# Data hadling
from sklearn.preprocessing import LabelEncoder
LabelEncoder = LabelEncoder()
dataset['Sex'] = LabelEncoder.fit_transform(dataset['Sex'])
dataset['Cabin'] = LabelEncoder.fit_transform(dataset['Cabin'])
dataset['Embarked'] = LabelEncoder.fit_transform(dataset['Embarked'])
```

```python
dataset.head(3)
```

```python
dataset.isnull().sum()
```

```python
dataset['Age'].fillna(dataset['Age'].mean(), inplace=True)
```

```python
y = dataset['Survived']
dataset = dataset.drop('Survived', axis=1)
```

```python
dataset.head(3)
```

```python
X = dataset.iloc[:,:].values
```

```python
X.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)
```

```python
StandardScaler = StandardScaler()
X_train = StandardScaler.fit_transform(X_train)
X_test = StandardScaler.fit_transform(X_test)
```

# Neural Network

```python
model = Sequential()
model.add(Dense(5, input_dim=9, activation='relu'))
model.add(Dense(5, activation='relu'))
```

```python
# output layer
model.add(Dense(1, activation='sigmoid'))
```

```python
opt = SGD(learning_rate=0.01, momentum=0.9)
```

```python
model.compile(loss='binary_crossentropy', optimizer=opt, metrics=['accuracy'])
```

```python
print("Training...")
history = model.fit(X_train, y_train, epochs=150, batch_size=32, verbose=1)

loss, accuracy = model.evaluate(X_test, y_test)
print(f"\nAcurácia no teste: {accuracy*100:.2f}%")
```

Two intermidiam layers with 5 neuron each one   
Epoch: 150   
SGD(learning_rate=0.01, momentum=0.9): Stochastic Gradient Descent and momentun is to not stop over small holes

# SVM

```python
from sklearn.svm import SVC
svc = SVC() #gamma scale e kernel rbf
svc.fit(X_train, y_train)
svc.score(X_test, y_test)
```
